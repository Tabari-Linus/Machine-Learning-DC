```python
import tensorflow as tf
# from tensorflow.python.ops import rnn, rnn_cell
from tensorflow.keras.utils import to_categorical
import numpy as np
from sklearn.model_selection import train_test_split
mnist = './Data/mnist.npz'
```

```python
# Model Hyperparameters
time_steps = 28       # Rows in the image
input_size = 28       # Columns in each row (input features per step)
num_classes = 10      # Digits 0-9
batch_size = 64
epochs = 10
hidden_units = 128

# Define the RNN model
def build_rnn_model():
    model = tf.keras.Sequential([
        tf.keras.layers.SimpleRNN(hidden_units, input_shape=(time_steps, input_size), activation='tanh'),
        tf.keras.layers.Dropout(0.2),
        tf.keras.layers.Dense(num_classes, activation='softmax')
    ])
    return model
```

```python
# Load the data
with np.load(mnist) as data:
    x_train = data['x_train']
    y_train = data['y_train']
    x_test = data['x_test']
    y_test = data['y_test']

# Normalize the data
x_train = x_train / 255.0
x_test = x_test / 255.0

# One-hot encode the labels
y_train = to_categorical(y_train, num_classes=10)
y_test = to_categorical(y_test, num_classes=10)
# Compile the model
# Build the model 
model = build_rnn_model()

# Compile the model
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])


# Train the model
model.fit(x_train, y_train, batch_size=batch_size, epochs=epochs, validation_data=(x_test, y_test))

# Evaluate the model
test_loss, test_acc = model.evaluate(x_test, y_test)
print(f"Test accuracy: {test_acc:.4f}")
```
