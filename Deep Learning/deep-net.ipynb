```python
import tensorflow as tf
import numpy as np
from sklearn.model_selection import train_test_split
mnist = './Data/mnist.npz'
```

```python
# Hyperparameters
batch_size = 100
input_size = 784  # For flattened 28x28 images
num_classes = 10  # Digits 0-9

# Build the neural network model
def neural_network_model():
    model = tf.keras.Sequential([
        tf.keras.layers.InputLayer(shape=(input_size,)),
        tf.keras.layers.Dense(128, activation='relu'),
        tf.keras.layers.Dense(128, activation='relu'),
        tf.keras.layers.Dense(num_classes, activation='softmax')
    ])
    return model

```

```python
# Load the data
with np.load(mnist) as data:
    x_train = data['x_train']
    y_train = data['y_train']
    x_test = data['x_test']
    y_test = data['y_test']

# Preprocess the data
x_train, x_test = x_train / 255.0, x_test / 255.0  # Normalize
x_train = x_train.reshape(-1, input_size)  # Flatten images
x_test = x_test.reshape(-1, input_size)
y_train = tf.keras.utils.to_categorical(y_train, num_classes)  # One-hot encode labels
y_test = tf.keras.utils.to_categorical(y_test, num_classes)

# Compile the model
model = neural_network_model()
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# Train the model
model.fit(x_train, y_train, epochs=10, batch_size=batch_size)

# Evaluate the model
test_loss, test_acc = model.evaluate(x_test, y_test)
print(f"Test accuracy: {test_acc:.4f}")
```
