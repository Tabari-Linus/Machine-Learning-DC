Pretrained Keras Model

```python
x = tf.random.normal(shape=(5, 299, 299, 3))
y = tf.constant([0,1,2,3,4])


model = keras.applications.InceptionV3(include_top=False,input_shape=(299, 299, 3))
# model.summary()

base_inputs = model.input
base_output = model.output
o = layers.GlobalAveragePooling2D()(base_output)
final_outputs = layers.Dense(5)(o)

new_model = keras.Model(inputs=base_inputs, outputs=final_outputs)

new_model.compile(
    loss=keras.losses.SparseCategoricalCrossentropy(from_logits=True),
    optimizer=keras.optimizers.Adam(),
    metrics=['accuracy']
)

new_model.fit(x, y, batch_size=1, epochs=15, verbose=2)

```
