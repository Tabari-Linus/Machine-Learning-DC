```python
import tensorflow as tf
from create_sentiment_featuresets import create_feature_sets_and_labels
```

```python
# Load the data
train_x, train_y, test_x, test_y = create_feature_sets_and_labels(
        'Data/pos.txt', 'Data/neg.txt'
    )

# Hyperparameters
batch_size = 100
input_size = len(train_x[0])  # For flattened 28x28 images
num_classes = 2  # Digits 0-9

# Build the neural network model
def neural_network_model():
    model = tf.keras.Sequential([
        tf.keras.layers.InputLayer(shape=(input_size,)),
        tf.keras.layers.Dense(128, activation='relu'),
        tf.keras.layers.Dense(128, activation='relu'),
        tf.keras.layers.Dense(num_classes, activation='softmax')
    ])
    return model



```

```python
# Preprocess the data
x_train, x_test = train_x,  test_x


# Compile the model
model = neural_network_model()
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# Train the model
model.fit(x_train, train_y, epochs=10, batch_size=batch_size)

# Evaluate the model
test_loss, test_acc = model.evaluate(x_test, test_y)
print(f"Test accuracy: {test_acc:.4f}")
```
