```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, regularizers
import numpy as np
mnist = './Data/mnist.npz'
```

```python
# Load the data
with np.load(mnist) as data:
    x_train = data['x_train']
    y_train = data['y_train']
    x_test = data['x_test']
    y_test = data['y_test']


x_train = x_train.astype('float32' )/ 255.0
x_test = x_test.astype('float32' )/ 255.0
```

```python
model = keras.Sequential()
model.add(keras.Input(shape=(None, 28)))
model.add(layers.SimpleRNN(512, return_sequences=True, activation='relu'))
model.add(layers.SimpleRNN(512, activation='relu'))
model.add(layers.Dense(10))


model.summary( )
```

```python
model.compile(
    loss=keras.losses.SparseCategoricalCrossentropy(from_logits=True),
    optimizer=keras.optimizers.Adam(0.001),
    metrics=['accuracy']
)

model.fit(x_train, y_train, batch_size=64, epochs=10, verbose=2)
model.evaluate(x_test,y_test, batch_size=64, verbose=2)
```

GRU model

```python
gru_model = keras.Sequential()
gru_model.add(keras.Input(shape=(None, 28)))
gru_model.add(layers.GRU(256, return_sequences=True, activation='relu'))
gru_model.add(layers.GRU(256, activation='relu'))
gru_model.add(layers.Dense(10))


gru_model.summary( )

gru_model.compile(
    loss=keras.losses.SparseCategoricalCrossentropy(from_logits=True),
    optimizer=keras.optimizers.Adam(0.001),
    metrics=['accuracy']
)

gru_model.fit(x_train, y_train, batch_size=64, epochs=10, verbose=2)
gru_model.evaluate(x_test,y_test, batch_size=64, verbose=2)
```

LSTM Model 

```python
lstm_model = keras.Sequential()
lstm_model.add(keras.Input(shape=(None, 28)))
lstm_model.add(layers.LSTM(512, return_sequences=True, activation='relu'))
lstm_model.add(layers.LSTM(512, activation='relu'))
lstm_model.add(layers.Dense(10))


lstm_model.summary( )

lstm_model.compile(
    loss=keras.losses.SparseCategoricalCrossentropy(from_logits=True),
    optimizer=keras.optimizers.Adam(0.001),
    metrics=['accuracy']
)

lstm_model.fit(x_train, y_train, batch_size=64, epochs=10, verbose=2)
lstm_model.evaluate(x_test,y_test, batch_size=64, verbose=2)
```

Bidirectionality in models 

```python
bidir_lstm_model = keras.Sequential()
bidir_lstm_model.add(keras.Input(shape=(None, 28)))
bidir_lstm_model.add(layers.Bidirectional(layers.LSTM(512, return_sequences=True, activation='relu')))
bidir_lstm_model.add(layers.Bidirectional(layers.LSTM(512, activation='relu')))
bidir_lstm_model.add(layers.Dense(10))


bidir_lstm_model.summary( )

bidir_lstm_model.compile(
    loss=keras.losses.SparseCategoricalCrossentropy(from_logits=True),
    optimizer=keras.optimizers.Adam(0.001),
    metrics=['accuracy']
)

bidir_lstm_model.fit(x_train, y_train, batch_size=64, epochs=10, verbose=2)
bidir_lstm_model.evaluate(x_test,y_test, batch_size=64, verbose=2)
```
